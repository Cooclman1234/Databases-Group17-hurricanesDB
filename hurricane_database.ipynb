Underneath, you find the necessary python methdos to populate the Database. You can add, delete and modify tuples of the DB.

```python
def add_hurricane(hurricane_id, name, start_datetime, end_datetime):
    sql = """
        INSERT INTO Hurricane (hurricane_id, name, start_datetime, end_datetime)
        VALUES (%s,%s,%s,%s)
    """
    cursor.execute(sql, (hurricane_id, name, start_datetime, end_datetime))
    connection.commit()
def add_location(location_id,region,country,latitude,longitude):
    sql = """
        INSERT INTO Location (location_id,region,country,latitude,longitude)
        VALUES (%s,%s,%s,%s,%s)
    """
    cursor.execute(sql, (location_id,region,country,latitude,longitude))
    connection.commit()
def add_meteorological_data(md_id, hurricane_id, location_id, start_datetime, end_datetime,avg_wind_speed, total_rainfall, avg_temperature):
    sql = """
        INSERT INTO MeteorologicalData (md_id, hurricane_id, location_id, start_datetime, end_datetime,avg_wind_speed, total_rainfall, avg_temperature)
        VALUES (%s,%s,%s,%s,%s,%s,%s,%s)
    """
    cursor.execute(sql, (md_id, hurricane_id, location_id, start_datetime, end_datetime,avg_wind_speed, total_rainfall, avg_temperature))
    connection.commit()
def add_destruction(destruction_id, md_id, deaths_direct, deaths_indirect, infrastructure_damage_usd, economic_damage_overall_usd):
    sql = """
        INSERT INTO Destruction (destruction_id, md_id, deaths_direct, deaths_indirect, infrastructure_damage_usd, economic_damage_overall_usd)
        VALUES (%s,%s,%s,%s,%s,%s)
    """
    cursor.execute(sql, (destruction_id, md_id, deaths_direct, deaths_indirect, infrastructure_damage_usd, economic_damage_overall_usd))
    connection.commit()

def delete_hurricane(hurricane_id):
    sql = """
        DELETE FROM Hurricane
        WHERE hurricane_id = %s;

    """
    cursor.execute(sql, (hurricane_id,))
    connection.commit()

def delete_location(location_id):
    sql = """
        DELETE FROM Location
        WHERE location_id = %s;

    """
    cursor.execute(sql, (location_id,))
    connection.commit()

def delete_meteorological_data(md_id):
    sql = """
        DELETE FROM MeteorologicalData
        WHERE md_id = %s;

    """
    cursor.execute(sql, (md_id,))
    connection.commit()

def delete_destruction(destruction_id):
    sql = """
        DELETE FROM Destruction
        WHERE destruction_id = %s;

    """
    cursor.execute(sql, (destruction_id,))
    connection.commit()

def update_hurricane(hurricane_id, name, start_datetime, end_datetime):
    sql = """
        UPDATE Hurricane
        SET name = %s,
            start_datetime = %s,
            end_datetime = %s
        WHERE hurricane_id = %s;
    """
    cursor.execute(sql, (name, start_datetime, end_datetime, hurricane_id))
    connection.commit()

def update_location(location_id, region, country, latitude, longitude):
    sql = """
        UPDATE Location
        SET region = %s,
            country = %s,
            latitude = %s,
            longitude = %s
        WHERE location_id = %s;
    """
    cursor.execute(sql, (region, country, latitude, longitude, location_id))
    connection.commit()

def update_meteorological_data(md_id, hurricane_id, location_id, start_datetime, end_datetime, avg_wind_speed, total_rainfall, avg_temperature):
    sql = """
        UPDATE MeteorologicalData
        SET hurricane_id = %s,
            location_id = %s,
            start_datetime = %s,
            end_datetime = %s,
            avg_wind_speed = %s,
            total_rainfall = %s,
            avg_temperature = %s
        WHERE md_id = %s;
    """
    cursor.execute(sql, (hurricane_id, location_id, start_datetime, end_datetime, avg_wind_speed, total_rainfall, avg_temperature, md_id))
    connection.commit()

def update_destruction(destruction_id, md_id, deaths_direct, deaths_indirect, infrastructure_damage_usd, economic_damage_overall_usd):
    sql = """
        UPDATE Destruction
        SET md_id = %s,
            deaths_direct = %s,
            deaths_indirect = %s,
            infrastructure_damage_usd = %s,
            economic_damage_overall_usd = %s
        WHERE destruction_id = %s;
    """
    cursor.execute(sql, (md_id, deaths_direct, deaths_indirect, infrastructure_damage_usd, economic_damage_overall_usd, destruction_id))
    connection.commit()
```

Code underneath creates the four realtions: Hurricane, Location, Destruction and Meterological_Data

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS Hurricane (
    hurricane_id INTEGER PRIMARY KEY AUTO_INCREMENT,
    name VARCHAR(120),
    start_datetime DATETIME,
    end_datetime DATETIME
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS Location (
    location_id INTEGER PRIMARY KEY AUTO_INCREMENT,
    region VARCHAR(120),
    country VARCHAR(120),
    latitude DECIMAL(9,6),
    longitude DECIMAL(9,6)
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS MeteorologicalData (
    md_id INTEGER PRIMARY KEY AUTO_INCREMENT,
    hurricane_id INTEGER,
    location_id INTEGER,
    start_datetime DATETIME,
    end_datetime DATETIME,
    avg_wind_speed DECIMAL(6,2),
    total_rainfall DECIMAL(7,2),
    avg_temperature DECIMAL(5,2),
    FOREIGN KEY (hurricane_id)
        REFERENCES Hurricane(hurricane_id),
    FOREIGN KEY (location_id)
        REFERENCES Location(location_id)
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS Destruction (
    destruction_id INTEGER PRIMARY KEY AUTO_INCREMENT,
    md_id INTEGER,
    deaths_direct INTEGER NOT NULL,
    deaths_indirect INTEGER NOT NULL,
    infrastructure_damage_usd BIGINT NOT NULL,
    economic_damage_overall_usd BIGINT,
    FOREIGN KEY (md_id)
        REFERENCES MeteorologicalData(md_id)
)
""")

connection.commit()

print("Tables created successfully.")
```

Undereath you find the advanced queries, we used...
methods:
get_specifc_hurricanes : fetches all hurricanes between the given time frame
get_meteorological_data_on_hurricane : fetches meterological data on a specific hurricane
get_sum_damage : calculates the complete damage of a location over the full lifetime

```python
def get_specific_hurricanes(start_datetime, end_datetime):
    query = """
    SELECT * FROM Hurricanes
    WHERE start_datetime >= ? AND end_datetime <= ?
    """
    cursor.execute(query, (start_datetime, end_datetime))
    return cursor.fetchall()

def get_meteorological_data_on_hurricane(hurricane_id):
    query = """
    SELECT * FROM MeteorologicalData
    WHERE hurricane_id = ?
    """
    cursor.execute(query, (hurricane_id,))
    return cursor.fetchall()

def get_sum_damage(location):
    sql = """
        SELECT Location.region, SUM(Destruction.deaths_direct), SUM(Destruction.deaths_indirect), SUM(Destruction.infrastructure_damage_usd), SUM(Destruction.economic_damage_overall_usd)
        FROM Location
        JOIN MeteorologicalData ON Location.location_id = MeteorologicalData.location_id
        JOIN Destruction ON MeteorologicalData.md_id = Destruction.md_id
        WHERE Location.region = %s
        GROUP BY Location.region;
    """
    cursor.execute(sql, (location,))
    return cursor.fetchone()

```
